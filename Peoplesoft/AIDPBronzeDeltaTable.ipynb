#### Read DB and write to AIDP standard table

```
def BronzeTablesFull(
    src_catalog: str,
    src_schema: str,
    tgt_catalog: str,
    tgt_schema: str,
    tbl_list: str
):

    tables_df = spark.sql(f"SHOW TABLES IN {src_catalog}.{src_schema}") \
        .filter("lower(tableName) <> 'dbtools_execution_history'")

    if tbl_list.strip() != "":
        tables_df = tables_df.filter(f"lower(tableName) in ({tbl_list})")

    for r in tables_df.collect():
        table_name = r["tableName"]
        full_name = f"{src_catalog}.{src_schema}.{table_name}"
        df = spark.read.table(full_name)

        tgt_name = f"{tgt_catalog}.{tgt_schema}.{table_name}"
        print(f"Target Table {tgt_name}")
        df.write.format("delta").mode("overwrite").saveAsTable(tgt_name)
        print(f"Table copied to={tgt_name}")
```
